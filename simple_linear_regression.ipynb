# Simple Linear Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset = pd.read_csv('Salary_Data.csv')
X = dataset.iloc[:, :-1].values
# iloc é usado para indexação baseada em inteiros.
# Seleciona todas as linhas (:) e todas as colunas, exceto a última (:-1).
# .values converte o resultado de um DataFrame Pandas em um array NumPy.
y = dataset.iloc[:, -1].values
# Seleciona todas as linhas (:) e apenas a última coluna (-1) do DataFrame.
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
# Essa função divide os dados em conjuntos de treinamento e teste
# para avaliação de modelos de aprendizado de máquina.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
# Divide os arrays X (variáveis independentes) e y (variável dependente) em quatro subconjuntos:
# X_train: dados de treinamento (features) usados para treinar o modelo.
# X_test: dados de teste (features) usados para avaliar o modelo.
# y_train: valores alvo correspondentes ao treinamento.
# y_test: valores alvo correspondentes ao teste.
# test_size=0.2:
# 20% dos dados serão usados para teste; os outros 80% serão para treinamento.
# random_state=0:
# Define uma semente fixa para o gerador aleatório, garantindo que a divisão seja reproduzível.
```

## Training the Simple Linear Regression model on the Training set

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
# Cria um objeto chamado regressor da classe LinearRegression.
# Esse objeto será usado para ajustar o modelo aos dados de treinamento e fazer previsões.
regressor.fit(X_train, y_train)
# fit(X_train, y_train):
# Treina o modelo de regressão linear nos dados de treinamento (X_train e y_train).
# O modelo aprende os coeficientes (inclinação da linha) e o intercepto (onde a linha cruza o eixo y).
```

## Predicting the Test set results

```python
y_pred = regressor.predict(X_test)
# predict(X_test):
# Usa o modelo treinado para prever os valores da variável dependente (y) para os dados de teste (X_test).
# y_pred agora contém os valores de salário previstos para o conjunto de teste.
```

## Visualising the Training set results

```python
plt.scatter(X_train, y_train, color = 'red')
# Cria um gráfico de dispersão para os dados de treinamento.
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salário vs Experiência (Conjunto de Treinamento)')
plt.xlabel('Anos de Experiência')
plt.ylabel('Salário')
plt.show()
```

## Visualising the Test set results

```python
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salário vs Experiência (Conjunto de Teste)')
plt.xlabel('Anos de Experiência')
plt.ylabel('Salário')
plt.show()
```
